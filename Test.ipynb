# TeLLAgent

TeLLAgent is a two-agent framework powered by tool-enhanced LLMs, designed to autonomously orchestrate the end-to-end workflow of organic material design and discovery. The framework comprises a global planning agent  for query reasoning and sub-task specification, and a local execution agent that precisely invokes domain-specific tools to accomplish complex tasks. 

## Here, we have shown how to use TeLLAgent for the discovery of organic materials
